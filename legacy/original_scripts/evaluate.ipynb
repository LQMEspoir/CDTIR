```python
#有真实值

import json
import re

# 加载现有的JSON数据
file_path = 'Qwen2.5_predicted_data.json'  # 输入文件路径
with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

# 处理数据，将predicted_output转换为有效的JSON对象列表
for entry in data:
    # 获取predicted_output字段
    predicted_output_str = entry.get("predicted_output", "")
    
    # 如果predicted_output字段有数据
    if predicted_output_str:
        # 使用正则表达式提取每个 JSON 对象
        predicted_output = re.findall(r'\{.*?\}', predicted_output_str)
        
        # 将每个提取出来的字符串转换为JSON对象
        predicted_output = [json.loads(item) for item in predicted_output]
        
        # 更新predicted_output字段
        entry["predicted_output"] = predicted_output

# 处理数据，将output转换为有效的JSON对象列表
for entry in data:
    # 获取output字段
    output_str = entry.get("output", "")
    
    # 如果output字段有数据
    if output_str:
        # 使用正则表达式提取每个 JSON 对象
        output = re.findall(r'\{.*?\}', output_str)
        
        # 将每个提取出来的字符串转换为JSON对象
        output = [json.loads(item) for item in output]
        
        # 更新output字段
        entry["output"] = output

# 将处理后的数据保存到新文件
output_path = 'processed_data_Qwen2.5.json'
with open(output_path, 'w', encoding='utf-8') as output_file:
    json.dump(data, output_file, ensure_ascii=False, indent=4)

print(f"处理后的数据已保存到 {output_path}")
```

```python
#加入extra_label和output_label标签
#生成文件：processed_data_with_label.json

import json

def process_and_format_json(input_file, output_file):
    formatted_data = []  # 用于存储标准格式的输出数据

    with open(input_file, 'r', encoding='utf-8') as f:
        # 读取整个文件
        data = json.load(f)

        for item in data:
            # 从输入数据中提取字段
            instruction = item.get("instruction", "")
            input_value = item.get("input", "")
            predicted_output = item.get("predicted_output", [])
            output = item.get("output", [])

            # 创建额外标签字段
            extra_labels = {ent: [] for ent in ["出发地","目的地", "返程地", "餐饮", "住宿", "产品", "交通", "预算", "时长", "时间", "人群", "强度", "人气", "人数", "天气"]}
            output_labels = {ent: [] for ent in ["出发地","目的地", "返程地", "餐饮", "住宿", "产品", "交通", "预算", "时长", "时间", "人群", "强度", "人气", "人数", "天气"]}

            # 将预测和输出中的实体分类到相应的标签字段
            for entity in predicted_output:
                entity_label = entity.get("entity_label", "")
                if entity_label in extra_labels:
                    extra_labels[entity_label].append(entity["entity_text"])

            for entity in output:
                entity_label = entity.get("entity_label", "")
                if entity_label in output_labels:
                    output_labels[entity_label].append(entity["entity_text"])

            # 添加 extra_label 和 output_label 字段
            item["extra_label"] = extra_labels
            item["output_label"] = output_labels

            # 添加到结果列表
            formatted_data.append(item)

    # 将格式化后的数据写入输出文件
    with open(output_file, 'w', encoding='utf-8') as out_f:
        json.dump(formatted_data, out_f, ensure_ascii=False, indent=2)  # 美化输出并保存

# 使用函数进行格式转换
input_file = 'processed_data_Qwen2.5.json'  # 输入文件路径
output_file = 'processed_data_Qwen2.5_with_label.json'  # 输出文件路径
process_and_format_json(input_file, output_file)
```

# 准确率、召回率和F1评估
精确率：识别出正确的实体数 / 识别出的实体数

召回率：识别出正确的实体数 / 样本的实体数

F1值 = (精确率 * 召回率 * 2) / ( 精确率 + 召回率)

## 代码核心思路
将预测结果与label转为集合，再利用集合的与操作，即可判断出模型预测成功的实体；

Node:
predict_right_num：当前文本，模型预测正确的实体数；

predict_num：模型预测实体总数；

label_num：label中真实的实体数；

```python
from dataclasses import dataclass

@dataclass
class Node:
    # 默认值
    predict_right_num: int = 0
    predict_num: int = 0
    label_num: int = 0
```

```python
import json

# 实体类别
ent_class = ["出发地", "目的地", "返程地", "餐饮", "住宿", "产品", "交通", "预算", "时长", "时间", "人群", "强度", "人气", "人数", "天气"]

class Node:
    def __init__(self):
        self.predict_right_num = 0  # TP: 预测正确的数量
        self.predict_num = 0        # TP + FP: 预测的总数量
        self.label_num = 0          # TP + FN: 实际标签的数量
        self.tn_num = 0             # TN: 预测为负，且标签也为负

def compute(input_file):
    with open(input_file, 'r', encoding='utf-8') as f:
        total_ent = {ent: Node() for ent in ent_class}
        error = 0
        
        data = json.load(f)

        for item in data:
            extra_labels = item.get("extra_label", {})
            output_label = item.get("output_label", {})

            for ent_name in ent_class:
                extra_s = set(extra_labels.get(ent_name, []))
                predict_s = set(output_label.get(ent_name, []))

                node = total_ent[ent_name]
                node.predict_right_num += len(extra_s & predict_s)
                node.predict_num += len(predict_s)
                node.label_num += len(extra_s)

                # 如果该类别在预测和标签中都不存在（属于 TN）
                if len(extra_s) == 0 and len(predict_s) == 0:
                    node.tn_num += 1

    print("每个实体类别的评估结果如下：\n")

    for ent in ent_class:
        node = total_ent[ent]
        tp = node.predict_right_num
        fp = node.predict_num - tp
        fn = node.label_num - tp
        tn = node.tn_num

        # 若 predict_num 和 label_num 都为 0，说明模型和标签都判断为无该类，是一次正确的负判断
        if node.predict_num == 0 and node.label_num == 0:
            precision = 1.0
            recall = 1.0
            f1 = 1.0
            acc = 1.0
        else:
            precision = tp / (tp + fp + 1e-6)
            recall = tp / (tp + fn + 1e-6)
            f1 = 0 if precision + recall == 0 else 2 * precision * recall / (precision + recall)
            acc = (tp + tn) / (tp + fp + fn + tn + 1e-6)

        print(f"{ent} -> precision: {precision:.4f}  recall: {recall:.4f}  f1: {f1:.4f}  acc: {acc:.4f}")
compute('processed_data_Qwen2.5_with_label.json')
```

```python
import json

def compute_using_entities(input_file):
    total_ent = {
        "correct": 0,    # TP：预测正确的实体总数
        "predicted": 0,  # TP + FP：模型预测的实体总数
        "actual": 0,     # TP + FN：真实标签的实体总数
    }

    with open(input_file, 'r', encoding='utf-8') as f:
        data = json.load(f)

        for item in data:
            extra_labels = item.get("extra_label", {})
            output_labels = item.get("output_label", {})

            # 统一处理文本格式：去除空格，转小写
            extra_entities = set()
            for ent in extra_labels:
                for entity in extra_labels[ent]:
                    extra_entities.add(entity.strip().lower())

            output_entities = set()
            for ent in output_labels:
                for entity in output_labels[ent]:
                    output_entities.add(entity.strip().lower())

            correct_entities = extra_entities & output_entities

            total_ent["correct"] += len(correct_entities)
            total_ent["predicted"] += len(output_entities)
            total_ent["actual"] += len(extra_entities)

    # 获取最终统计数
    correct = total_ent["correct"]
    predicted = total_ent["predicted"]
    actual = total_ent["actual"]

    # 特殊处理：完全没预测、也没标签的情况，视为完美预测
    if predicted == 0 and actual == 0:
        precision = 1.0
        recall = 1.0
        f1 = 1.0
    else:
        precision = correct / (predicted + 1e-6)
        recall = correct / (actual + 1e-6)
        f1 = 2 * precision * recall / (precision + recall + 1e-6)

    # 输出评估结果
    print("【整体实体匹配评估】")
    print(f"总预测实体数（Predicted）: {predicted}")
    print(f"总真实实体数（Actual）   : {actual}")
    print(f"预测正确实体数（Correct）: {correct}")
    print(f"整体精确率 Precision     : {precision:.4f}")
    print(f"整体召回率 Recall        : {recall:.4f}")
    print(f"整体F1值 F1-score        : {f1:.4f}")
compute_using_entities('processed_data_Qwen2.5_with_label.json')
```
